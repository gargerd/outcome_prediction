```python
import pandas as pd
import numpy as np
import os
import time
import pickle
from matplotlib import pyplot as plt
import itertools
import seaborn as sns

os.getcwd()
os.chdir(os.getcwd())
os.getcwd()
```

```python
phase1_df=pd.read_csv('../data/phase_1_df.csv',low_memory=False,index_col=0)
```

### Plot some laboratory variables over time

```python
pat_df=phase1_df.loc['TB-1007/1069',:].dropna(how='all',axis=1)
pat_df['mb_Colony Count_STD_NUM_RESULT']=np.log(pat_df['mb_Colony Count_STD_NUM_RESULT'].replace(0,1))
numerical_cols=pat_df.columns[pat_df.columns.str.contains('STD_NUM_RESULT_SCALED$')].tolist()
pat_df.loc[:,numerical_cols]
pat_df=pat_df.set_index('DAY',drop=True)
plot_df=pat_df.copy()

## PLot some lab variables
num_of_vars=30
fig,ax=plt.subplots(1,1,figsize=(17,12))
with sns.axes_style("whitegrid"):
    a=sns.lineplot(data=plot_df.loc[:,numerical_cols[0:num_of_vars]],ax=ax,markers=False)
    b=sns.scatterplot(data=plot_df.loc[:,numerical_cols[0:num_of_vars]],ax=ax,s=75,label=None)
    a.set_yticklabels(np.round(a.get_yticks(),1), size = 15)
    a.set_xticklabels(a.get_xticks(), size = 15)
    ax.set_xlabel('Study day',fontsize=25,fontweight='bold')
    ax.set_ylabel('Normalised values',fontsize=25,fontweight='bold')
    handles, labels = ax.get_legend_handles_labels()
    labels_new=[x.split('_')[1] for x in labels]
    ax.legend(handles, labels_new[0:num_of_vars],loc='upper right')
    ax.set_title('Laboratory variables over time',fontsize=30,fontweight='bold')

## Plot MGIT TTP over time
num_of_vars=1
fig,ax=plt.subplots(1,1,figsize=(17,12))
with sns.axes_style("darkgrid"):
    sns.lineplot(data=pat_df,x='DAY',y='mb_Time to Detection_STD_NUM_RESULT',ax=ax)
    a=sns.scatterplot(data=pat_df,x='DAY',y='mb_Time to Detection_STD_NUM_RESULT',ax=ax,s=75,label=None)
    ax.set_ylim(0,)
    a.set_yticklabels(np.round(a.get_yticks(),1), size = 15)
    a.set_xticklabels(a.get_xticks(), size = 15)
    ax.set_xlabel('Study day',fontsize=25,fontweight='bold')
    ax.set_ylabel('TTP (Days)',fontsize=25,fontweight='bold')
    handles, labels = ax.get_legend_handles_labels()
    labels_new=[x.split('_')[1] for x in labels]
    ax.legend(handles, labels_new[0:num_of_vars],loc='upper right')
    ax.set_title('MGIT TTP values in first 2 weeks therapy',fontsize=30,fontweight='bold')
```
