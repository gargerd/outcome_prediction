## Define LSTM

```python
# Define the LSTM model
class LSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, output_size,device):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Sequential(nn.Linear(hidden_size, int(hidden_size/2)),
                                nn.ReLU(),
                                nn.Linear(int(hidden_size/2),int(hidden_size/4)),
                                nn.ReLU(),
                                nn.Linear(int(hidden_size/4),output_size))
        self.device=device
        
    def forward(self, x):
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(self.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(self.device)
        
        out, _ = self.lstm(x)
        out = self.fc(out)
        out=out.reshape(-1)
        return out
```

## 1. __Preprocess data for ML and split into X and y matrices__
*  Drop columns with strings that are unnecessary and create one-hot encoded columns of categorical variables
*  Split data into training variables (X) and variable to predict (y== MGIT TTP values)
*  Split data into training variables (X) and variable to predict (log(y): log MGIT TTP values))

## 2. __Function to split data into train and test data and fit models__

-  1. Method: train on data containing MGIT measurements from whole study range -> predict MGIT measurements at any timepoint
-  2. Method: train on the data of first X days of therapy -> predict MGIT measurements of therapy days > X


## 3. Train different predicitive models ans save the result of prediction
* #### Split data into train and test data
* #### Train multiple predictive methods (RandomForest, linear regression, Ridge regression, Lasso regression, Elastic net)
* #### Predict MGIT TTP using test data and save the results
* #### Cross-validation of the training of the multiple models

```python
### Drop unnecessary columns from the whole dataset, one-hot encode categorical variables and return X and y dataframes 
def create_X_y_dataframe_from_raw_data(df_for_analysis,y_as_log):
    ## Drop unnecessary columns and create one-hot encoding of categorical vars
    data=df_for_analysis.drop(columns=['USUBJID','STUDYID','ARM'])

    ## Convert the ordinal categorical columns to numbers from string
    #  Select columns with string values &  columns with only 2 unique values
    ordinal_categorical_cols=data.select_dtypes('object').columns[data.select_dtypes('object').columns.str.contains('_')&\
                                                                  data.select_dtypes('object').apply(lambda x: len(x.unique())==2)]

    ## Dictionary containing the variables to replace with numerical values
    ordinal_cat_replace_dict={'|'.join(['RESISTANT','Y','YES']):1,
                              '|'.join(['SUSCEPTIBLE','N','NO']):0}

    ## Replace categorical values with ordered numerical values
    #data.loc[:,ordinal_categorical_cols]=data.loc[:,ordinal_categorical_cols].replace(ordinal_cat_replace_dict,regex=True)
    
    ## Select nominal columns with categorical variables >3 unique values and convert them to one-hot encoded variables
    data_hot_encoded=pd.get_dummies(data)

    ## Drop microbiological measurement columns (except for TTP measurements) + Study DAY information, 
    #  as these are not needed for prediction
    #mb_cols=data_hot_encoded.columns[data_hot_encoded.columns.str.contains('mb_',na=False)].tolist()
    #cols_to_drop=mb_cols #+ ['DAY']
    
    ## If microbiological test result that is considered for analysis, don't drop it/them
    #if len(mb_cols_to_keep)>0:
    #    cols_to_drop=cols_to_drop.remove(mb_cols_to_keep)

    ## Drop target variable from predictor variables' dataframe
    cols_to_drop=['mb_Time to Detection_STD_NUM_RESULT']
    X=data_hot_encoded.drop(columns=cols_to_drop)


    ## Create y data to predict + keep some data (day, study id, arm) for plotting later
    y=df_for_analysis[['DAY','STUDYID','ARM','mb_Time to Detection_STD_NUM_RESULT']]
    y=y.set_index(['DAY','STUDYID','ARM'],drop=True)
    y_log=np.log10(y.replace(0,1))

    if y_as_log==False:
        return X,y
    if y_as_log==True:
        return X,y_log

        

## 1. Method: train on whole dataset and predict MGIT TTP at any timepoint
def training_on_whole_data(data,data_split_random_state,y_as_log):

    X,y=create_X_y_dataframe_from_raw_data(data,y_as_log)
    ## Split training data
    test_data_ratio=0.25

    ## Standardise numerical columns
    num_cols=X.select_dtypes(exclude=['object']).columns.tolist()
    X.loc[:,num_cols]=(StandardScaler().fit_transform(X.loc[:,num_cols].values))

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_data_ratio, random_state=data_split_random_state,\
                                                        shuffle=True)
    return (X_train, X_test, y_train, y_test)



## 2. Method: train only on first X days and predict MGIT TTP values after therapy day X
def training_on_first_x_days(training_days,data,y_as_log):
    
    ## Split training data by study days -> train only on the first X days, and predict the TTP values for the days > X
    X,y=create_X_y_dataframe_from_raw_data(data,y_as_log)
    X_train=X[X['DAY']<=training_days]
    X_test=X[X['DAY']>training_days]
    y_train=y[y.index.get_level_values('DAY')<=training_days]
    y_test=y[y.index.get_level_values('DAY')>training_days]
    
    ## Standardise training data's numerical columns
    for X in [X_train,X_test]:
        num_cols=X.select_dtypes(exclude=['object']).columns.tolist()
        X.loc[:,num_cols]=(StandardScaler().fit_transform(X.loc[:,num_cols].values))
        
    return (X_train, X_test, y_train, y_test)



## Split the training data with the selected method and transform 'y' dataframes to colum vector necessary for ML algorithms
def split_training_data(training_method,data,training_days,data_split_random_state):
     ## 1. Method
    if training_method=='training_on_whole_data':
        X_train, X_test, y_train, y_test=training_on_whole_data(data,data_split_random_state,\
                                                                y_as_log=False)

        X_train, X_test, y_train_log, y_test_log=training_on_whole_data(data,data_split_random_state,\
                                                                y_as_log=True)
    ## 2. Method
    if training_method=='training_on_first_x_days':
        X_train, X_test, y_train, y_test=training_on_first_x_days(training_days,data,\
                                                                  y_as_log=False)
        X_train, X_test, y_train_log, y_test_log=training_on_first_x_days(training_days,data,\
                                                                          y_as_log=True)

    ## Extract the y values from y_test dataframes in order to feed it to ML algorithms
    #  (The multiindex of y_test dataframes will be used later for coloring the plots)
    y_test_vals=y_test.values.ravel()
    y_test_log_vals=y_test_log.values.ravel()
    y_train_vals=y_train.values.ravel()
    y_train_log_vals=y_train_log.values.ravel()

    return(X_train, X_test, y_train, y_test,y_train_log, y_test_log,y_test_vals,y_test_log_vals,y_train_vals,y_train_log_vals)




## Function to train different prediction models. Save results of prediction if needed with cross-validation
def train_models(training_method,data,model_names,scoring_methods,train_dataset_types,training_days,
                 random_state,data_split_random_state,test_data_results,cv_results,lstm_parameters,cross_validation,\
                 get_feature_importances):
    

    ## Add training method to the result dictionary
    test_data_results[training_method]={}

    ## Split training and testing dataset based on the method chosen
    X_train, X_test, y_train, y_test,y_train_log, y_test_log,\
        y_test_vals,y_test_log_vals,y_train_vals,y_train_log_vals=split_training_data(training_method,data,\
                                                                                      training_days,data_split_random_state)
                                                                                      
    ## Collect the train and test data types( raw and logarithmic) into lists to loop over during training and prediction
    y_train_datasets=[y_train_vals,y_train_log_vals]
    y_test_datasets=[y_test_vals,y_test_log_vals]
    # Make list of y_test dataframes where the MultiIndex contains information for plotting (i.e. Day, Studyid, etc..)
    y_train_datasets_with_index=[y_train,y_train_log]
    y_test_datasets_with_index=[y_test,y_test_log]

    ## If CV true, then save results of CV and show them
    if cross_validation==True:
        cv_results[training_method]={}
        

    ## Loop over methods and train the models. If CV==true, perform CV and plot the results
    for train_ds_type,y_train_data,y_test_data_vals,y_test_data_with_index,y_train_data_with_index in zip(train_dataset_types,y_train_datasets,y_test_datasets,\
                                                                                   y_test_datasets_with_index,y_train_datasets_with_index):

        for model_type_name in model_names:
            model_name=model_type_name+'_'+train_ds_type

            if 'LSTM' in model_name:
                # Convert the data to PyTorch tensors
                X_train_tensor = torch.tensor(X_train.values, dtype=torch.float32)
                y_train_data_tensor = torch.tensor(y_train_data, dtype=torch.float32)

                # Initialize the model and optimizer
                device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
                model=LSTM(X_train_tensor.shape[1], lstm_parameters['hidden_size'], \
                            lstm_parameters['num_layers'], lstm_parameters['output_size'],device).to(device)
                criterion=lstm_parameters['criterion'] #
                optimizer=torch.optim.Adam(model.parameters(), lr=lstm_parameters['learning_rate'])

                # Train the model
                num_of_epochs=lstm_parameters['num_epochs']
                print('X_train_tensor shape',X_train_tensor.shape)
                for epoch in range(num_of_epochs):
                    outputs = model(X_train_tensor.to(device))
                    loss = criterion(outputs, y_train_data_tensor.to(device))
                    
                    optimizer.zero_grad()
                    loss.backward()
                    optimizer.step()
                    
                    if (epoch+1) % 10 == 0:
                        print(f'Epoch [{epoch+1}/{num_of_epochs}], Loss: {loss.item():.4f}')

                # Evaluate the model
                with torch.no_grad():
                    X_test_tensor = torch.tensor(X_test.values, dtype=torch.float32).to(device)
                    predicted_output = model(X_test_tensor).cpu().numpy()
                    actual_output = y_test_data_vals.reshape(-1, 1)
                    print(f'Test MSE: {mean_squared_error(actual_output, predicted_output):.4f}')
                    test_data_results[training_method][model_name+'_y']=y_test_data_vals
                    test_data_results[training_method][model_name+'_y_with_index']=y_test_data_with_index
                    test_data_results[training_method][model_name+'_train_y_with_index']=y_train_data_with_index
                    test_data_results[training_method][model_name+'_predicted']=predicted_output
                #feature_names=model.feature_names_in_
                #feature_importances=model.feature_importances_

        if training_method=='training_on_first_x_days':
                test_data_results[training_method]['training_days']=(training_days)

    if cross_validation==False:
        return  test_data_results                   
```

```python
parameters_for_analysis={'phase1_391_pats_59_vars':{'selection_method':'phase',
                            'phase':'1',
                            'graph_metric':None,
                            'num_of_common_vars':59,
                            'training_days':10},

                        'phase2_920_pats_22_vars':{'selection_method':'phase',
                            'phase':'2',
                            'graph_metric':None,
                            'num_of_common_vars':22,
                            'training_days':45},

                        'phase3_1648_pats_42_vars':{'selection_method':'phase',
                            'phase':'3',
                            'graph_metric':None,
                            'num_of_common_vars':42,
                            'training_days':120},
                        
                        'pat_graph_679_pats_49_vars':{'selection_method':'patient_graph',
                            'phase':'1-2',
                            'graph_metric':'Page rank',
                            'num_of_common_vars':49,
                            'training_days':30},

                        'pat_graph_390_pats_53_vars':{'selection_method':'patient_graph',
                            'phase':'2',
                            'graph_metric':'Page rank',
                            'num_of_common_vars':53,
                            'training_days':30},


                        'var_graph_519_pats_60_vars':{'selection_method':'variable_graph',
                            'phase':'2',
                            'graph_metric':'Page rank',
                            'num_of_common_vars':60,
                            'training_days':30},

                        'var_graph_2246_pats_29_vars':{'selection_method':'variable_graph',
                            'phase':'1-2-3',
                            'graph_metric':'Page rank',
                            'num_of_common_vars':29,
                            'training_days':120}}
```

## Run LSTM

```python
## Choose CV features: number of folds and repeats for CV in the training + scoring method
k_folds=5
num_of_cv_repeats=1
scoring_methods=['neg_mean_squared_error','r2']

## List of training dataset types: raw MGIT TTP values or logaritmic MGIT TTP values
train_dataset_types=['raw','log']

## List of prediction models
model_names=['LSTM']


## Set training method and number of days to train on if training method is 'training_on_first_x_days'
training_methods=['training_on_whole_data','training_on_first_x_days']
random_state=44
data_split_random_state=44

## List of temporal dataframe names to drop from loaded dataframe (as they are highly sparse, perhaps only add noise?)
temporal_data_names=['ae_','cmind_','cmdos_','cmday_','mh']
temporal_data_to_drop=['cmdos']
# If True, drop columns of temporal data given in the 'temporal_data_to_drop' list
drop_temporal=True

## Boolean if we want to update exsiting result_dict -> if False, the training will be run on only not-yet existing 
#  +ds_name+'_preproc_data.csv' files
update_existing_dicts=True

for ds_name in [*parameters_for_analysis][0:]:

    ## Select the number of training days for the training method 'training_on_first_x_days'
    parameter_dict=parameters_for_analysis[ds_name]
    training_days=parameter_dict['training_days']
    print('dataset name: ',ds_name)

    ## Load the preprocessed + imputed dataset
    data=pd.read_csv('../data/'+ds_name+'_preproc_data.csv',index_col=0) 

    ## Drop predefined temporal data columns 
    data=data.loc[:,~data.columns.str.startswith(tuple(temporal_data_to_drop))]
    data=data.set_index('index')

    
    ## Extract the name of the retained temporal data (i.e. 'mh','ae') -> add them as a suffix to 'dict_name' ->
    #  later it can be traced back which temporal columns were used in the analysis
    retained_temp_data_names=data.columns[data.columns.str.startswith(tuple(temporal_data_names))]
    retained_temp_data_names=list(set([x.split('_')[0] for x in retained_temp_data_names]))
    ds_name_with_temp_df_names='_'.join([ds_name] +retained_temp_data_names)
    

    ## Check if result_dict already exists. If yes, load it, if not create a new one
    fname='../data/'+ds_name+'_LSTM_results.pickle'
    if os.path.exists(fname):
        with open(fname,'rb') as f:
            result_dict=pickle.load(f)
    
    ## If result_dict doesn't exist yet or we want to update the existing result_dict -> 
    #  create empty result_dict and collect the results in it
    if os.path.exists(fname)==False or update_existing_dicts==True:
        result_dict={}
        result_dict[ds_name_with_temp_df_names]={}

    ## Check if test_data_results or cv_results already exists -> if not create them
    for result_type in ['test_data_results','cv_results','feat_importances']:
        if result_type not in result_dict[ds_name_with_temp_df_names].keys():
            result_dict[ds_name_with_temp_df_names][result_type]={}
    
    # Define the hyperparameters
    lstm_parameters={'input_size' :len(data.columns),
                'hidden_size':64,
                'num_layers':1,
                'output_size':1,
                'learning_rate':0.01,
                'num_epochs':40,
                'criterion':nn.MSELoss()}

    print('Shape of final data: ',data.shape) 
    print()

    ## Create dicts to collect test data, CV-results and importance scores for all training methods
    test_data_results={}
    cv_results={}
    importances_dict={}
    
    for training_method in training_methods[:]:
        ### TRAIN MODELS AND SAVE CV-RESULTS + TEST DATA RESULTS #####
        ## Check if data already exists for given dataset -> only run training when training hasn't been run yet

        if training_method not in result_dict[ds_name_with_temp_df_names]['test_data_results'].keys() \
            or training_method not in result_dict[ds_name_with_temp_df_names]['cv_results'].keys():

            ## Update previously created dicts with ML results
            test_data_results=train_models(training_method,data,model_names,scoring_methods,\
                                                    train_dataset_types,training_days,random_state,data_split_random_state,\
                                                    test_data_results,cv_results,lstm_parameters,cross_validation=False,\
                                                    get_feature_importances=False)
            

            ## Add results of test data and CV to the result_dict
            result_dict[ds_name_with_temp_df_names]['test_data_results']=test_data_results 
            #result_dict[ds_name_with_temp_df_names]['cv_results']=cv_results  

    ## Save results of ML on the given data
    with open('../data/'+ds_name+'_LSTM_results.pickle', 'wb') as f:
        pickle.dump(result_dict, f)            
```

```python
import warnings
warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)

figtitle_size=30
subplot_title_size=16
ylabel_size=16
text_size=18
legend_title_fontsize=13


model_names=['LSTM']
#train_dataset_types=['raw','log']

color_by='DAY'
color_by_legend_title={'STUDYID':'Study','DAY':'Study day','ARM':'Arm'}
training_method_fig_title={'training_on_whole_data':'Trained on all study days - Predicting TTP on all study days',
                           'training_on_first_x_days':'Trained on up until day'}

for ds_name in [*parameters_for_analysis][:]:
    
    ## Load the preprocessed + imputed dataset
    data=pd.read_csv('../data/'+ds_name+'_preproc_data.csv',index_col=0)   

    fname='../data/'+ds_name+'_LSTM_results.pickle'
    if os.path.exists(fname):
        with open(fname,'rb') as f:
            result_dict=pickle.load(f)
    

    for ds_name_with_temp_df_names in [*result_dict][:]:
        training_data_results=result_dict[ds_name_with_temp_df_names]['test_data_results']
        training_days=training_data_results['training_on_first_x_days']['training_days']

        for training_method in training_methods[:]:
            #X_train, X_test, y_train, y_test,y_train_log, y_test_log,\
            #y_test_vals,y_test_log_vals,y_train_vals,y_train_log_vals=split_training_data(training_method,data,\
            #                                                                              training_days,data_split_random_state)
            

            ## Select items from dict with acti=ual and predicted values that were saved as np.arrays
            plot_dict={key: value for key, value in training_data_results[training_method].items() if '_with_index' not in key}
            pred_df=pd.DataFrame.from_dict(plot_dict)
            
            fig,axes=plt.subplots(len(train_dataset_types),len(model_names),figsize=(len(train_dataset_types)*6,len(model_names)*16))

            for coln,model_type_name in enumerate(model_names[:]):
                for rown,train_ds_type in enumerate(train_dataset_types):
                    #if 'log' in train_ds_type:
                    #    pred_df=pred_df.apply(lambda x: 10**x)
                    if len(model_names)==1:
                        ax=axes[rown]
                    if len(train_dataset_types)==1:
                        ax=axes[coln]     
                    if len(model_names)>1 and len(train_dataset_types)>1:
                        ax=axes[rown,coln]
                    model_name=model_type_name+'_'+train_ds_type
                    actual_y=model_name+'_y'
                    actual_y_with_index=model_name+'_y_with_index'
                    y_train_name=model_name+'_train_y_with_index'
                    predicted=model_name+'_predicted'

                    y_test= training_data_results[training_method][actual_y_with_index]
                    y_train=training_data_results[training_method][y_train_name]
                    study_length=int(max(y_test.index.get_level_values('DAY').max(),y_train.index.get_level_values('DAY').max()))
                    prediction_duration=range(training_days+1,study_length+1)

                    max_value=pred_df[[actual_y,predicted]].max().max()
                    min_value=pred_df[[actual_y,predicted]].min().min()
                    r2=round(r2_score(pred_df[actual_y], pred_df[predicted]),2)
                    mse=round(sklearn.metrics.mean_squared_error(pred_df[actual_y], pred_df[predicted]),2)

                    hue=y_test.index.get_level_values(color_by)
                    plot_df=pred_df.copy()
                    plot_df[color_by]=hue.astype(float).tolist()
                    sns.scatterplot(data=plot_df,x=actual_y,y=predicted,ax=ax,hue=color_by,s=15,\
                                    vmin=plot_df[color_by].min(),vmax=50)#legend='auto')
                    sns.regplot(data=plot_df,x=actual_y,y=predicted,ax=ax,scatter=False)
                    ax.set_ylim(min_value,max_value*1.1)
                    ax.set_xlim(min_value,max_value*1.1)
                    text=r"$R^2$: "+ str(r2)+'\nMSE: '+str(mse)
                    axis_range=max_value*1.1-min_value
                    ax.text(0.5*axis_range+min_value,0.82*axis_range+min_value,text,fontsize=text_size)

                    ylabel='Predicted '+ train_ds_type+' TTP (Days)'
                    ax.set_ylabel(ylabel,fontsize=ylabel_size,fontweight='bold')
                    xlabel='Actual '+ train_ds_type+' TTP (Days)'
                    ax.set_xlabel(xlabel,fontsize=ylabel_size,fontweight='bold')

                    ax.plot(ax.get_xlim(), ax.get_ylim(),color='black', linestyle='dashed')
                    ax.set_title(model_name,fontsize=subplot_title_size,fontweight='bold')

                    ax.legend(title=color_by_legend_title[color_by],title_fontsize=legend_title_fontsize)

            if training_method=='training_on_first_x_days':
                suptitle=ds_name_with_temp_df_names.replace('_',' ')+'\n'+training_method_fig_title[training_method] +\
                         ' '+ str(training_days)+' - Predicting TTP on days '+str(prediction_duration[0])+ ' - '+str(prediction_duration[-1]) 
                         
            else:
                suptitle=ds_name_with_temp_df_names.replace('_',' ')+'\n '+training_method_fig_title[training_method] 
            fig.suptitle(suptitle,fontsize=figtitle_size,fontweight='bold')
            plt.tight_layout()


```
