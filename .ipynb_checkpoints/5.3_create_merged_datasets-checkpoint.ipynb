#### **1. MERGE DATA FROM DIFFERENT DATASETS (i.e. mb,dm,pc...) INTO ONE DATAFRAME FOR EACH PATIENT**
##### FORMAT:
*  COLUMNS: VARNAME_COLUMNNAME (COLUMNNAMES FROM AX_NAMES DICT)
*  INDEX/ROWS: TIMEPOINT OF MEASUREMENT (DAY)

#### **2. SAVE EVERY MERGED DATASET OR EACH PATIENT INTO ONE DICTIONARY (merged_ds_dict)**

```python
# put datasets that are considered for analysis into one list for iteration 
ds_dicts=[ttp_pc_dict,ttp_mb_dict,ttp_vs_dict,ttp_lb_blood_dict,ttp_lb_urine_dict,ttp_mol_res_dict,ttp_mic_dict,ttp_ms_dict]
ds_names=['ttp_pc_dict','ttp_mb_dict','ttp_vs_dict','ttp_lb_blood_dict','ttp_lb_urine_dict','ttp_mol_res_dict','ttp_mic_dict','ttp_ms_dict']

#if multiple numerical measurements from one day, get their daily means and return as a dataframe
def get_daily_mean_of_numeric(df,ds_dict_name):
    n_res_name=ax_names[ds_dict_name]['num_result']
    cat_res_name=ax_names[ds_dict_name]['cat_result']
    if df[n_res_name].isna().all()==False:
        col=[e for e in df.columns.tolist() if e not in [n_res_name,cat_res_name]]
        df=df.groupby(by=col).mean().reset_index()
    return df

def merge_datasets_into_dict(pat_ids,ds_dicts,ds_names,vars_for_analysis,ttp_temporal_pat_regimens):
    num=0
    num_allpat=len(pat_ids)
    start = time.time()

    merged_ds_dict={}
    for id in pat_ids:
        num=num+1
        print(id)
        print(num/num_allpat*100,'%')
        ds_df_list=[]
        for ds_dict,ds_dict_name in zip(ds_dicts,ds_names):
            try:
                pat_ds_dict=ds_dict[id]
            except KeyError:
                continue
            var_df_list=[]

            # select only variables in the dtaframe that is considered for this analysis (drop the variables that don't have anough timepoints)
            ds_vars_for_analysis=list(set(pat_ds_dict['variables'])&set(vars_for_analysis))
            
            # Only create dataframe of a variable if it is contained in the list of considered variables
            if len(ds_vars_for_analysis)>0:
                #iterate through the selected variables of one patient->create a df with results for one variable and append them to var_df_list
                for var in ds_vars_for_analysis:
                    df=pat_ds_dict['grouped_data'].get_group(var)
                    df=df.drop_duplicates()
                    df=df.loc[:,ax_names[ds_dict_name]['col_names']]

                    #check if results are numerical -> then get the daily means of the numerical measurements
                    if 'num_result' in ax_names[ds_dict_name].keys():
                        df=get_daily_mean_of_numeric(df,ds_dict_name)
                    #to identify which column name is referrring to which variable, concatenate 'var_' in the front of column names from ax_names dict 
                    #(i.e. var=pc_Bedaquiline, PCTEST->pc_Bedaquiline_PCTEST)
                    df=df.set_index(keys=ax_names[ds_dict_name]['index_names'],drop=True)
                    update_cols=[var+'_'+x for x in df.columns]
                    df.columns=update_cols
                    var_df_list.append(df)
                # initialize var_df with same multiindex created in previuos step
                n=len(ax_names[ds_dict_name]['index_names'])
                mult_ind=pd.MultiIndex(levels=[[]]*n,codes=[[]]*n,names=ax_names[ds_dict_name]['index_names'])
                var_df=pd.DataFrame(index=mult_ind)
                #var_df: df containing results of all the variables of ONE DATASET PER PATIENT
                for d in var_df_list:
                    var_df=pd.merge(var_df,d,left_index=True,right_index=True,how='outer')
                var_df.index.rename('DAY',level=0,inplace=True)
                ds_df_list.append(var_df)

            #initialize pat_df:
            n=len(common_index)
            mult_ind=pd.MultiIndex(levels=[[]]*n,codes=[[]]*n,names=common_index)
            pat_df=pd.DataFrame(index=mult_ind)
            #pat_df: df containing results of all the variables of MULTIPLE DATASETS PER PATIENT
            for ds in ds_df_list:
                pat_df=pd.merge(pat_df,ds,left_index=True,right_index=True,how='outer')

            #subset the temporal drug regimen of patient from ttp_temporal_pat_regimens and merge to pat_df
            reg_df=ttp_temporal_pat_regimens[ttp_temporal_pat_regimens['USUBJID']==id]
            reg_df=reg_df.set_index(keys=common_index,drop=True)
            pat_df=pd.merge(pat_df,reg_df,left_index=True,right_index=True,how='outer')

            #add the demographic data, which has constant values to the patients dataframe
            dm_dict=ttp_dm_dict[id]
            dm_colnames=ax_names['ttp_dm_dict']['col_names']
            pat_df.loc[:,dm_colnames]=dm_dict['bulk_data'][dm_colnames].values[0]

            merged_ds_dict[id]=pat_df


    end = time.time()
    print((end - start)/60,'minutes runtime')
    return merged_ds_dict


#pat_df
```
