### Get start and end day of the adverse events
* #### AIM: where necessary impute __START DAY__ and __END DAY__ of adverse event
* #### __Save the dataset where start and end day of adverse event is available__



```python
#len(d)
```

```python
### Initiate standardised column names
ae['STD_AEDY']=ae['AEDY']
ae['STD_AESTDY']=ae['AESTDY']
ae['STD_AEENDY']=ae['AEENDY']
ae['STD_AEDUR']=ae['AEDUR']
ae=ae.drop_duplicates()

#--------------------------------
#### Drop measurement point without any temporal information
#### + Drop measurement points originating from more than 20 days before starting day of therapy
ae_with_time=ae[~((ae['AEDY'].isna())&(ae['AESTDY'].isna())&(ae['AEENDY'].isna()))]
ae_with_time=ae_with_time[((ae_with_time['AEDY']>0)|(ae_with_time['AEENDY']>0)|(ae_with_time['AESTDY']>0))]

#--------------------------------
#### Select datapoints, where start day is missing, but can be calculated from end day and duration:
#### start day = end day - duration of adverse event

d=ae_with_time[(ae_with_time['AESTDY'].isna())&(~ae_with_time['AEDUR'].isna())&
             (~ae_with_time['AEENDY'].isna())]
if len(d)>0 :            
    ae_with_time.loc[d.index,'STD_AESTDY']=d['AEENDY']-d['AEDUR'].str.split(pat=r'[P,D]',expand=True)[1].astype(float)
ae_with_time=ae_with_time.drop_duplicates()
#--------------------------------
### Subset dataframe to points where either start or end day or both are missing
ae_missing_start_or_end_day=ae_with_time[~((~ae_with_time['STD_AESTDY'].isna())&(~ae_with_time['STD_AEENDY'].isna()))]

#### Select datapoints, where start day and end day are missing, but can be calculated from day of visit and duration:
#### start day = visit day - duration of adverse event
#### end day= visit day

d=ae_missing_start_or_end_day[(ae_missing_start_or_end_day['AESTDY'].isna())&(~ae_missing_start_or_end_day['AEDUR'].isna())&
             (ae_missing_start_or_end_day['AEENDY'].isna())]
if len(d)>0 : 
    ae_with_time.loc[d.index,'STD_AESTDY']=d['AEDY']-d['AEDUR'].str.split(pat=r'[P,D]',expand=True)[1].astype(float)
    ae_with_time.loc[d.index,'STD_AEENDY']=d['AEDY']

#--------------------------------
### Where day of visit and end day of AE event are equal and Start day and Duration is missing,
### assume that AE lasted only one day -> Start day of AE is equal to Day of Visit
d=ae_missing_start_or_end_day[(ae_missing_start_or_end_day['AESTDY'].isna())&(ae_missing_start_or_end_day['AEDUR'].isna())&
             ~(ae_missing_start_or_end_day['AEENDY'].isna())&~(ae_missing_start_or_end_day['AEDY'].isna())]
d=d[d['AEDY']==d['AEENDY']]
ae_with_time.loc[d.index,'STD_AEENDY']=d['AEDY']

#--------------------------------
### Where only day of visit is known, assume that AE occured only on that day
### Start day and End day of AE is equal to Day of Visit
d=ae_missing_start_or_end_day[(ae_missing_start_or_end_day['AESTDY'].isna())&(ae_missing_start_or_end_day['AEDUR'].isna())&
             (ae_missing_start_or_end_day['AEENDY'].isna())&~(ae_missing_start_or_end_day['AEDY'].isna())]
ae_with_time.loc[d.index,'STD_AESTDY']=d['AEDY']
ae_with_time.loc[d.index,'STD_AEENDY']=d['AEDY']

## Select data points where both start and end day of the events are available (drop data where severity is meaningless (==7))
ae_with_time=ae_with_time[((~ae_with_time['STD_AESTDY'].isna())&(~ae_with_time['STD_AEENDY'].isna()))]
#ae_with_time['AESEV']=ae_with_time['AESEV'].replace({'7':np.nan},regex=True)


## Select rows where information about the severity of the adverse event is available (either toxicity grade or severity)
ae_std=ae_with_time[~(ae_with_time['AETOXGR'].isna())]

## For some datapoint the toicity grade was set to 'A' -> assume that is equal to 1
ae_std['STD_AETOXGR']=ae_std['AETOXGR'].replace({'A':1},regex=True)


## Where toxicity grtade is missing, impute the grade from the severity information
#  Conversion: mild:1, moderate:2, severe:3, life threatening:4
#sev_list=['MILD','MODERATE','SEVERE']
#tox_grade_list=[1,2,3]
#sev_tox_grade_dict=dict(zip(sev_list,tox_grade_list))
#ae_std.loc[ae_std['STD_AETOXGR'].isna(),'STD_AETOXGR']=ae_std.loc[ae_std['STD_AETOXGR'].isna(),'AESEV'].map(sev_tox_grade_dict)
```

```python
## Initialise list to collect results into
ae_pat_df_list=[]

## Loop over all patients and extract the start and stop day,dose, application route for each drug applied over therapy period 
#  into temp_df then concatenate the saved temp_df of each patient to a final dataframe 
## For each drug create a column cm_drugname_application_route (i.e. cm_amoxicillin_oral) 
for pat_id in ae_std['USUBJID'].unique()[0:]:
    pat_df=ae_std[ae_std['USUBJID']==pat_id]
    temp_df=pd.DataFrame(columns=['DAY','STUDYID','USUBJID'],index=np.arange(pat_df['STD_AESTDY'].min(),pat_df['STD_AEENDY'].max()+1))

    temp_df['USUBJID']=pat_id
    temp_df['STUDYID']=pat_df['STUDYID'].values[0]
    temp_df['DAY']=temp_df.index.tolist()

    ## Loop over each adverse event of patient
    for _,event_df in pat_df.groupby(by='STD_AETERM'):

        ## If there are mulitple rows in event_df, that means patient may have had same adverse event multiple times
        #  -> loop over rows to catch all this information
        for row in event_df.index:
            adv_event=event_df.loc[row,'STD_AETERM']
            event_colname='_'.join(['ae',adv_event])

            ## Create column for event if it doesn't exist yet
            if event_colname not in temp_df.columns.tolist():
                temp_df[event_colname]=np.nan

            tox_grade=event_df.loc[row,'STD_AETOXGR']
            start=event_df.loc[row,'STD_AESTDY']
            stop=event_df.loc[row,'STD_AEENDY'] 

            temp_df.loc[start:stop,event_colname]=tox_grade

    ## Drop rows with only NaNs
    nan_rows_mask=~(temp_df.loc[:,temp_df.columns.str.contains('ae',na=False)].isna().all(1))
    temp_df=temp_df[nan_rows_mask]

    ## Drop rows with all NaNs and append temp_df to final list
    ae_pat_df_list.append(temp_df)

## Concatenate all the dataframes for each patients
ae_temporal_with_toxicity=pd.concat(ae_pat_df_list,ignore_index=True)
```
